```python
data = pd.read_csv("SEM_scaled_clean_data.csv")

y = "salary_06"
T = "select"
X = ['dwomen', 'select_dwomen', 'age_lb', 'empl_04', 'salary_04',
    'hours_04', 'days_04', 'contract_04', 'dformal_04', 'pempl_04',
    'dmarried_lb', 'city_2_0', 'city_3_0', 'city_4_0', 'city_5_0', 'city_6_0',
    'city_1_0', 'educ_lb_2_0', 'educ_lb_3_0', 'educ_lb_4_0', 'educ_lb_5_0',
    'educ_lb_6_0', 'educ_lb_7_0', 'educ_lb_8_0', 'educ_lb_9_0', 'educ_lb_10_0',
    'educ_lb_0_0', 'educ_lb_12_0', 'educ_lb_13_0', 'educ_lb_14_0', 'educ_lb_16_0']
X_mat = data[X].values
t_vec = data[T].values
y_vec = data[y].values
```

## S-Learner

### ATE

```python
learner_s = BaseSRegressor(XGBRegressor(), control_name=0)
ate_s = learner_s.estimate_ate(X=X_mat, treatment=t_vec, y=y_vec, return_ci=True, 
                               bootstrap_ci=True, n_bootstraps=100)
```

```python
print("S-learner ATE (point, CI):", ate_s)
```

### CATE

```python
alpha = 0.05
learner_s = BaseSRegressor(XGBRegressor(), ate_alpha=alpha, control_name=0)
cate_s, cate_s_lb, cate_s_ub = learner_s.fit_predict(X=X_mat, treatment=t_vec, y=y_vec, return_ci=True,
                               n_bootstraps=100, bootstrap_size=5000)
```

```python
data = data.assign(cate_s=cate_s, cate_s_lb=cate_s_lb, cate_s_ub=cate_s_ub)

print("CATE_s (first 5):")
for i in range(5):
    val  = cate_s[i].item()       # or cate_s[i][0]
    low  = cate_s_lb[i].item()
    high = cate_s_ub[i].item()
    print(f"{val:.4f} [{low:.4f}, {high:.4f}]")
```

## T-Learner

### ATE w/ Boostrap Confidence Intervals

```python
learner_t = BaseTRegressor(XGBRegressor(), control_name=0)
ate_t = learner_t.estimate_ate(X=X_mat, treatment=t_vec, y=y_vec, bootstrap_ci=True, n_bootstraps=100)
```

```python
print("T-learner ATE (point, CI):", ate_t)
```

### CATE w/ Confidence Intervals

```python
learner_t = BaseTRegressor(XGBRegressor(), control_name=0)
cate_t, cate_t_lb, cate_t_ub = learner_t.fit_predict(X=X_mat, treatment=t_vec, y=y_vec, return_ci=True, n_bootstraps=100,
                                                    bootstrap_size=5000)
```

```python
data = data.assign(cate_t=cate_t, cate_t_lb=cate_t_lb, cate_t_ub=cate_t_ub)

print("CATE_t (first 5):")
for i in range(5):
    val  = cate_t[i].item()       # or cate_s[i][0]
    low  = cate_t_lb[i].item()
    high = cate_t_ub[i].item()
    print(f"{val:.4f} [{low:.4f}, {high:.4f}]")
```

## Visualize

```python
groups = learner_r._classes

alpha = 1
linewidth = 2
bins = 30
for group,idx in sorted(groups.items(), key=lambda x: x[1]):
    plt.figure(figsize=(12,8))
    plt.hist(cate_t[:,idx], alpha=alpha, bins=bins, label='T Learner ({})'.format(group),
             histtype='step', linewidth=linewidth, density=True)
    plt.hist(cate_x[:,idx], alpha=alpha, bins=bins, label='X Learner ({})'.format(group),
             histtype='step', linewidth=linewidth, density=True)
    plt.hist(cate_r[:,idx], alpha=alpha, bins=bins, label='R Learner ({})'.format(group),
             histtype='step', linewidth=linewidth, density=True)
    plt.hist(tau, alpha=alpha, bins=bins, label='Actual ATE distr',
             histtype='step', linewidth=linewidth, color='green', density=True)
    plt.vlines(cate_s[0,idx], 0, plt.axes().get_ylim()[1], label='S Learner ({})'.format(group),
               linestyles='dotted', linewidth=linewidth)
    plt.vlines(tau.mean(), 0, plt.axes().get_ylim()[1], label='Actual ATE',
               linestyles='dotted', linewidth=linewidth, color='green')
    
    plt.title('Distribution of CATE Predictions for {}'.format(group))
    plt.xlabel('Individual Treatment Effect (ITE/CATE)')
    plt.ylabel('# of Samples')
    _=plt.legend()
```
